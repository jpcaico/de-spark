```python
from pyspark.sql import SparkSession, functions as func
```

```python
spark = SparkSession.builder.appName("Customers").getOrCreate()
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType
```

```python
schema = StructType([
StructField("customerID", StringType(), True),
StructField("itemID", StringType(), True),
StructField("amount", FloatType(), True)
])
```

```python
customers = spark.read.schema(schema).csv("/opt/spark/data/customer-orders.csv")
```

```python
customers.show(5)
```

```python
results = customers.groupBy("customerID").sum("amount").sort("customerID")
```

```python
results_rounded = customers.groupBy("customerID").agg(func.round(func.sum("amount"), 2).alias("total_amount")).sort("customerID")
```
