```python
from pyspark import SparkConf, SparkContext
import collections
```

```python
conf = SparkConf().setMaster("local").setAppName("RatingsHistogram")
sc = SparkContext(conf = conf)
```

```python
## Load data
# Breaks input file line by line. Each line is one line in the RDD
lines = sc.textFile("/opt/spark/data/ml-100k/u.data")
```

```python
print(lines)
```

```python
# Extract (map) the data we care about
ratings = lines.map(lambda x: x.split()[2])
```

```python
ratings
```

```python
# Perform an action: count by value

results = ratings.countByValue()
```

```python
results
```

```python
# sort and display results

sortedResults = collections.OrderedDict(sorted(results.items()))
```

```python
sortedResults
```

```python
for key, value in sortedResults.items():
    print( (key, value) )
```
