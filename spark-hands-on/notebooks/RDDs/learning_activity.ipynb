```python
# Add up amount spent by customer
# id, item_id, amount_spent

# split each comma-delimited line into fields
# map each line to key/value pairs of customer id and dollar amount
# use reduceByKey to add up amount spent by customer id
# collect() the results and print them
```

```python
from pyspark import SparkConf, SparkContext

conf = SparkConf().setMaster("local").setAppName("CustomerAmounts")
```

```python
sc = SparkContext(conf=conf)
```

```python
lines = sc.textFile("/opt/spark/data/customer-orders.csv")
```

```python
def parseLines(line):
    fields = line.split(',')
    customerID = fields[0]
    amount_spent = fields[2]
    return (int(customerID), float(amount_spent))
    
```

```python
parsedLines = lines.map(parseLines)
```

```python
parsedLines
```

```python
amount = parsedLines.reduceByKey(lambda x,y: x + y)
amountSorted = amount.map(lambda pair: (pair[0], pair[1])).sortByKey()
```

```python
results = amountSorted.collect()
```

```python
for result in results:
    print(result)
```
