Spark can do special stuff with key/value data:

- `reduceByKey()`: combine values with the same key using some function. `rdd.reduceByKey(lambda x, y: x + y)` adds them up
- `groupByKey()`: Group values with the same key
- `sortByKey()`: Sort RDD by key values
- `keys()` `values()` : Create an RDD of just the keys or just the values

With key/value data, use `mapValues()` and `flatMapValues()` if your transformation doesn't affect the keys. It's more efficient.

```python
from pyspark import SparkConf, SparkContext

conf = SparkConf().setMaster("local").setAppName("FriendsByAge")
sc = SparkContext(conf = conf)
```

```python
def parseLine(line):
    fields = line.split(',')
    age = int(fields[2])
    numFriends = int(fields[3])
    return (age, numFriends)
```

```python
lines = sc.textFile("/opt/spark/data/fakefriends.csv")
rdd = lines.map(parseLine)
```

```python
# count up sum of friends and number of entries per age
totalsByAge = rdd.mapValues(lambda x: (x,1)).reduceByKey(lambda x,y: (x[0] + y[0], x[1] + y[1]))
```

```python
# compute averages
averagesByAge = totalsByAge.mapValues(lambda x: x[0] / x[1])
```

```python
# collect and show results
results = averagesByAge.collect()
```

```python
for result in results:
    print(result)
```
